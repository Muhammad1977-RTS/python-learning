```python
import sqlite3
import pandas as pd

conn = sqlite3.connect('game_events.db')

with open('create_game_events.txt', 'r', encoding='utf-8') as f:
    sql_script = f.read()

conn.executescript(sql_script)

df = pd.read_sql('SELECT * FROM game_events', conn)
conn.close()

print(df)
```

```python
df.to_csv('game_events.csv', index=False)
```
